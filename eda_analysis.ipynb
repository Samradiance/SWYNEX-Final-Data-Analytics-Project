```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_music_dataset.csv")
df.head()
```

```python
print("number of rows:",df.shape[0])
print("number of columns:",df.shape[1])
```

```python
print("Column Name:")
for column in df.columns:
    print(column)
```

```python
df.info
```

```python
df.describe()
```

```python
## Missining Value Analysis

missing_values = df.isnull().sum()
print("Missing values in each columns:")
print(missing_values)
```

```python
## Duplicate Record Analysis

duplicate_count = df.duplicated().sum()
print("Number of duplicate rows: ",duplicate_count)
```

```python
## Genre Distributon

genre_counts = df["genre"].value_counts()
print(genre_counts)
```

```python
genre_counts.head(10).plot(kind="bar", figsize=(10,6))
plt.title("Top 10 Genres by Number of Tracks")
plt.xlabel("Genre")
plt.ylabel("Number of Tracks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Stream Count Analysis

print("Average streams: ",df["stream_count"].mean())
print("Median streams: ",df["stream_count"].median())
print("Minimum streams: ",df["stream_count"].min())
print("Maximum streams: ",df["stream_count"].max())
```

```python
# Stream Distribution

plt.figure(figsize=(10,6))

plt.hist(df["stream_count"] ,bins=50)

plt.title("Distribution of Stream Counts")
plt.xlabel("Stream Count")
plt.xlabel("Number of Tracks")

plt.tight_layout()
plt.show()
```

```python
# Top 10 Most Streamed Tracks

top_tracks = df.nlargest(10, "stream_count")
top_tracks[["track_name", "artist_name", "genre", "stream_count"]]
```

```python
## Release Year Analysis

year_counts = df["release_year"].value_counts().sort_index()
print(year_counts)
```

```python
year_counts.plot(kind="line", marker="o", figsize=(10,6))

plt.title("Number of Tracks Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Tracks")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
## Average Streams by Year

average_streams_year = df.groupby("release_year")["stream_count"].mean()
print(average_streams_year)
```

```python
average_streams_year.plot(kind="line", marker="o", figsize=(10,6))
plt.title("Average Streams by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Average Stream Count")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
# Genre vs Average Streams

genre_streams = df.groupby("genre")["stream_count"].mean().sort_values(ascending=False)
print(genre_streams)
```

```python
genre_streams.plot(kind="bar", figsize=(12,6))
plt.title("Average Streams by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Stream Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Popularity Analysis

print(df["popularity"].describe())
```

```python
plt.figure(figsize=(10,6))
plt.hist(df["popularity"], bins=20)
plt.title("Distribution of Music Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Tracks")
plt.tight_layout()
plt.show()
```

```python
# Popularity vs Streams

plt.figure(figsize=(10,6))
plt.scatter(
    df["popularity"],
    df["stream_count"],
    alpha=0.3
)
plt.title("Popularity vs Stream Count")
plt.xlabel("Popularity")
plt.ylabel("Stream Count")
plt.tight_layout()
plt.show()
```

```python
correlation = df["popularity"].corr(df["stream_count"])
print("Correlation between popularity and stream count: ", correlation)
```

```python
# Audio Feature Analysis

audio_features =[
    "danceability",
    "energy",
    "instrumentalness",
    "tempo",
    "loudness",
    "upbeat_score"
]

df[audio_features].describe()
```

```python
## Audio features Correlation with Streams

correlation = df[
    audio_features + ["stream_count"]
].corr()["stream_count"].sort_values(ascending=False)
print(correlation)
```

```python
## Explict Music Analysis

explicit_analysis = df.groupby("is_explicit_bool").agg(
    tracks=("track_id", "count"),
    average_streams=("stream_count", "mean"),
    average_popularity = ("popularity", "mean")
)
explicit_analysis
```
